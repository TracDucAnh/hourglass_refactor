# Study 2a — bộ sinh ngôn ngữ tổng hợp và kết quả

Mã ở `src/hourglass/study2_synthetic/`. Cổng G0: `make g0`; huấn luyện: `scripts/sweep.sh`.

```python
from hourglass.study2_synthetic import load

ws = load("study2a", gpu=False)
```

```python
# cùng một nội dung C, ba mức khác biệt M0 / M2 / M4 (overlap âm vị 0%)
C = ws.ContentSampler("demo").sample(False)
for lev in ws.LEVELS:
    for s in ws.make_languages(3, lev, 0, [1.0] * 3, seed=7):
        print(f"{lev} {s.describe():70s} | {ws.render(s, C)}")
```
